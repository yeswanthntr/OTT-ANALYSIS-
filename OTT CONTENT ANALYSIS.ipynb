```python
# ============================================================
# OTT STREAMING CONTENT TREND ANALYSIS
# COMPLETE EDA CODE
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully")


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(r"/content/sample_data/netflix_titles.csv")

print("Dataset loaded successfully")

print("Dataset shape:", df.shape)


# ============================================================
# 3. FIRST LOOK AT DATA
# ============================================================

print("\nFirst 5 rows:")
print(df.head())


print("\nLast 5 rows:")
print(df.tail())


print("\nColumn names:")
print(df.columns.tolist())


# ============================================================
# 4. DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
print(df.info())


print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# ============================================================
# 5. STATISTICAL SUMMARY
# ============================================================

print("\nStatistical Summary:")
print(df.describe(include="all"))


# ============================================================
# 6. CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate rows:", duplicate_count)


# Remove duplicates

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)


# ============================================================
# 7. MISSING VALUE ANALYSIS
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# Missing percentage

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)


missing_table = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})


missing_table = missing_table[
    missing_table["Missing Count"] > 0
].sort_values(
    "Missing Percentage",
    ascending=False
)


print("\nMissing Value Summary:")
print(missing_table)


# ============================================================
# 8. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nCleaned Column Names:")
print(df.columns.tolist())


# ============================================================
# 9. DATA TYPE CONVERSION
# ============================================================

# Convert date_added to datetime

df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)


# Convert release_year to numeric

df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
)


print("\nData types after conversion:")
print(df.dtypes)


# ============================================================
# 10. HANDLE MISSING VALUES
# ============================================================

# Fill categorical missing values

df["director"] = df["director"].fillna("Unknown")

df["cast"] = df["cast"].fillna("Unknown")

df["country"] = df["country"].fillna("Unknown")

df["rating"] = df["rating"].fillna("Unknown")

df["duration"] = df["duration"].fillna("Unknown")


# ============================================================
# 11. CREATE YEAR ADDED COLUMN
# ============================================================

df["year_added"] = df["date_added"].dt.year


print("\nDate information:")
print(
    df[
        ["date_added", "year_added", "release_year"]
    ].head()
)


# ============================================================
# 12. BASIC CONTENT TYPE ANALYSIS
# ============================================================

type_count = df["type"].value_counts()

print("\nMovies vs TV Shows:")
print(type_count)


# Percentage

type_percentage = (
    df["type"]
    .value_counts(normalize=True)
    * 100
).round(2)


print("\nMovies vs TV Shows Percentage:")
print(type_percentage)


# ============================================================
# 13. VISUALIZATION 1
# MOVIES VS TV SHOWS
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="type"
)

plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.tight_layout()
plt.show()


# ============================================================
# 14. RELEASE YEAR ANALYSIS
# ============================================================

year_count = (
    df["release_year"]
    .value_counts()
    .sort_index()
)


print("\nContent by Release Year:")
print(year_count.tail(20))


# ============================================================
# 15. VISUALIZATION 2
# RELEASE YEAR TREND
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    year_count.index,
    year_count.values,
    marker="o"
)

plt.title("Content Release Trend by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 16. GENRE ANALYSIS
# ============================================================

# Fill missing genres

df["listed_in"] = df["listed_in"].fillna("Unknown")


# Split multiple genres

genre_df = df.assign(
    genre=df["listed_in"].str.split(", ")
).explode("genre")


print("\nGenre data:")
print(
    genre_df[
        ["title", "genre"]
    ].head(10)
)


# Genre count

genre_count = (
    genre_df["genre"]
    .value_counts()
)


print("\nTop 10 Genres:")
print(
    genre_count.head(10)
)


# ============================================================
# 17. VISUALIZATION 3
# TOP 10 GENRES
# ============================================================

top_genres = genre_count.head(10)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title("Top 10 Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.tight_layout()
plt.show()


# ============================================================
# 18. COUNTRY ANALYSIS
# ============================================================

df["country"] = df["country"].fillna("Unknown")


country_df = df.assign(
    country=df["country"].str.split(", ")
).explode("country")


country_count = (
    country_df["country"]
    .value_counts()
)


print("\nTop 10 Countries:")
print(
    country_count.head(10)
)


# ============================================================
# 19. VISUALIZATION 4
# TOP 10 COUNTRIES
# ============================================================

top_countries = country_count.head(10)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title(
    "Top 10 Countries by Number of OTT Titles"
)

plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.tight_layout()
plt.show()


# ============================================================
# 20. RATING ANALYSIS
# ============================================================

df["rating"] = df["rating"].fillna("Unknown")


rating_count = (
    df["rating"]
    .value_counts()
)


print("\nRating Distribution:")
print(rating_count)


# ============================================================
# 21. VISUALIZATION 5
# RATING DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=rating_count.values,
    y=rating_count.index
)

plt.title("Content Rating Distribution")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()


# ============================================================
# 22. RATING BY CONTENT TYPE
# ============================================================

rating_type = pd.crosstab(
    df["rating"],
    df["type"]
)


print("\nRating by Content Type:")
print(rating_type)


# ============================================================
# 23. VISUALIZATION 6
# RATINGS BY CONTENT TYPE
# ============================================================

top_ratings = (
    df["rating"]
    .value_counts()
    .head(8)
    .index
)


rating_type_top = pd.crosstab(
    df[
        df["rating"].isin(top_ratings)
    ]["rating"],
    df[
        df["rating"].isin(top_ratings)
    ]["type"]
)


rating_type_top.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "Ratings by Content Type"
)

plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 24. MOVIE DURATION ANALYSIS
# ============================================================

# Extract number from duration

df["duration_minutes"] = (
    df["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)


# Select movies

movies = df[
    df["type"] == "Movie"
].copy()


print("\nNumber of Movies:", len(movies))


# Basic statistics

print("\nMovie Duration Statistics:")
print(
    movies["duration_minutes"].describe()
)


# Average movie duration

print(
    "\nAverage movie duration:",
    round(
        movies["duration_minutes"].mean(),
        2
    ),
    "minutes"
)


# Shortest movie

if movies["duration_minutes"].notna().any():

    shortest_index = (
        movies["duration_minutes"]
        .idxmin()
    )

    print("\nShortest movie:")

    print(
        movies.loc[
            shortest_index,
            [
                "title",
                "duration_minutes"
            ]
        ]
    )


# Longest movie

if movies["duration_minutes"].notna().any():

    longest_index = (
        movies["duration_minutes"]
        .idxmax()
    )

    print("\nLongest movie:")

    print(
        movies.loc[
            longest_index,
            [
                "title",
                "duration_minutes"
            ]
        ]
    )


# ============================================================
# 25. VISUALIZATION 7
# MOVIE DURATION DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    movies["duration_minutes"].dropna(),
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Movie Durations"
)

plt.xlabel(
    "Duration (Minutes)"
)

plt.ylabel(
    "Number of Movies"
)

plt.tight_layout()
plt.show()


# ============================================================
# 26. AVERAGE MOVIE DURATION BY YEAR
# ============================================================

avg_duration_year = (
    movies
    .groupby("release_year")[
        "duration_minutes"
    ]
    .mean()
    .dropna()
)


# ============================================================
# 27. VISUALIZATION 8
# AVERAGE MOVIE DURATION BY YEAR
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    avg_duration_year.index,
    avg_duration_year.values,
    marker="o"
)

plt.title(
    "Average Movie Duration by Release Year"
)

plt.xlabel(
    "Release Year"
)

plt.ylabel(
    "Average Duration (Minutes)"
)

plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 28. TV SHOW SEASON ANALYSIS
# ============================================================

tv_shows = df[
    df["type"] == "TV Show"
].copy()


print(
    "\nNumber of TV Shows:",
    len(tv_shows)
)


# Extract number of seasons

tv_shows["seasons"] = (
    tv_shows["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)


print("\nTV Show Season Data:")

print(
    tv_shows[
        [
            "title",
            "duration",
            "seasons"
        ]
    ].head(10)
)


# Season distribution

season_count = (
    tv_shows["seasons"]
    .value_counts()
    .sort_index()
)


print(
    "\nTV Show Season Distribution:"
)

print(season_count)


# Average seasons

average_seasons = (
    tv_shows["seasons"].mean()
)


print(
    "\nAverage TV show seasons:",
    round(average_seasons, 2)
)


# ============================================================
# 29. TOP LONG-RUNNING TV SHOWS
# ============================================================

top_long_running = (
    tv_shows[
        [
            "title",
            "seasons"
        ]
    ]
    .sort_values(
        "seasons",
        ascending=False
    )
    .head(10)
)


print(
    "\nTop 10 Long-Running TV Shows:"
)

print(top_long_running)


# ============================================================
# 30. VISUALIZATION 9
# SEASON DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=season_count.index,
    y=season_count.values
)

plt.title(
    "Distribution of TV Shows by Number of Seasons"
)

plt.xlabel(
    "Number of Seasons"
)

plt.ylabel(
    "Number of TV Shows"
)

plt.tight_layout()
plt.show()


# ============================================================
# 31. AVERAGE SEASONS BY RELEASE YEAR
# ============================================================

avg_seasons_year = (
    tv_shows
    .groupby("release_year")[
        "seasons"
    ]
    .mean()
    .dropna()
)


# ============================================================
# 32. VISUALIZATION 10
# AVERAGE SEASONS BY YEAR
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    avg_seasons_year.index,
    avg_seasons_year.values,
    marker="o"
)

plt.title(
    "Average Number of Seasons by Release Year"
)

plt.xlabel(
    "Release Year"
)

plt.ylabel(
    "Average Number of Seasons"
)

plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 33. GENRE + RELEASE YEAR ANALYSIS
# ============================================================

genre_year = (
    genre_df
    .groupby(
        [
            "release_year",
            "genre"
        ]
    )
    .size()
    .reset_index(
        name="count"
    )
)


print(
    "\nGenre by Release Year:"
)

print(
    genre_year.head(10)
)


# ============================================================
# 34. TOP 5 GENRES
# ============================================================

top_5_genres = (
    genre_df["genre"]
    .value_counts()
    .head(5)
)


print(
    "\nTop 5 Genres:"
)

print(top_5_genres)


# ============================================================
# 35. CREATE GENRE TREND TABLE
# ============================================================

top_genre_names = (
    top_5_genres.index.tolist()
)


genre_trend = genre_year[
    genre_year["genre"].isin(
        top_genre_names
    )
]


genre_pivot = (
    genre_trend
    .pivot(
        index="release_year",
        columns="genre",
        values="count"
    )
    .fillna(0)
)


print(
    "\nGenre Trend Table:"
)

print(
    genre_pivot.tail(10)
)


# ============================================================
# 36. VISUALIZATION 11
# TOP 5 GENRE TRENDS
# ============================================================

plt.figure(figsize=(14, 7))

for genre in top_genre_names:

    plt.plot(
        genre_pivot.index,
        genre_pivot[genre],
        marker="o",
        label=genre
    )


plt.title(
    "Top 5 Genre Trends Over Release Years"
)

plt.xlabel(
    "Release Year"
)

plt.ylabel(
    "Number of Titles"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 37. RECENT 10-YEAR GENRE ANALYSIS
# ============================================================

genre_clean = genre_df.dropna(
    subset=["release_year"]
).copy()


genre_clean["release_year"] = (
    genre_clean["release_year"]
    .astype(int)
)


max_year = (
    genre_clean["release_year"].max()
)


recent_genres = genre_clean[
    genre_clean["release_year"]
    >= max_year - 9
]


recent_genre_count = (
    recent_genres["genre"]
    .value_counts()
    .head(10)
)


print(
    "\nTop Genres in Recent 10 Years:"
)

print(recent_genre_count)


# ============================================================
# 38. VISUALIZATION 12
# RECENT GENRE POPULARITY
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=recent_genre_count.values,
    y=recent_genre_count.index
)

plt.title(
    "Top Genres in the Most Recent 10 Years"
)

plt.xlabel(
    "Number of Titles"
)

plt.ylabel(
    "Genre"
)

plt.tight_layout()

plt.show()


# ============================================================
# 39. ONE-SEASON TV SHOW ANALYSIS
# ============================================================

one_season_percentage = (
    (
        tv_shows["seasons"] == 1
    ).mean()
    * 100
)


print(
    "\nTV shows with only one season:",
    round(
        one_season_percentage,
        2
    ),
    "%"
)


# ============================================================
# 40. FINAL SUMMARY
# ============================================================

summary = {

    "Total Titles":
        len(df),

    "Total Movies":
        (
            df["type"] == "Movie"
        ).sum(),

    "Total TV Shows":
        (
            df["type"] == "TV Show"
        ).sum(),

    "Total Genres":
        genre_df["genre"].nunique(),

    "Total Countries":
        country_df["country"].nunique(),

    "Total Ratings":
        df["rating"].nunique(),

    "Earliest Release Year":
        df["release_year"].min(),

    "Latest Release Year":
        df["release_year"].max(),

    "Average Movie Duration":
        round(
            movies[
                "duration_minutes"
            ].mean(),
            2
        ),

    "Average TV Show Seasons":
        round(
            tv_shows[
                "seasons"
            ].mean(),
            2
        ),

    "One Season TV Shows (%)":
        round(
            one_season_percentage,
            2
        )
}


summary_df = pd.DataFrame(
    summary.items(),
    columns=[
        "Metric",
        "Value"
    ]
)


print(
    "\n" + "=" * 60
)

print(
    "FINAL EDA SUMMARY"
)

print(
    "=" * 60
)

print(summary_df)


# ============================================================
# 41. SAVE CLEANED DATA
# ============================================================

df.to_csv(
    "cleaned_ott_data.csv",
    index=False
)


print(
    "\nCleaned dataset saved as:"
)

print(
    "cleaned_ott_data.csv"
)


# ============================================================
# 42. FINAL BUSINESS INSIGHTS
# ============================================================

print(
    "\n" + "=" * 60
)

print(
    "BUSINESS INSIGHTS"
)

print(
    "=" * 60
)


print("""
1. GENRE STRATEGY
The genre analysis identifies the most represented genres
and their trends over time. Streaming platforms can use
these trends to prioritize high-demand genres while
experimenting with emerging categories.


2. REGIONAL CONTENT STRATEGY
Content production is concentrated in major countries.
Increasing localized content from underrepresented regions
can help streaming platforms reach new audiences and
diversify their content portfolio.


3. MOVIE AND TV SHOW STRATEGY
The analysis of movies, TV shows, movie duration and TV show
seasons provides insights into the platform's content mix.
These findings can support decisions about short-form,
feature-length, limited-series and long-running content.
""")


# ============================================================
# 43. FINAL CONCLUSION
# ============================================================

print(
    "\n" + "=" * 60
)

print(
    "FINAL CONCLUSION"
)

print(
    "=" * 60
)

print("""
The exploratory data analysis provides a comprehensive view
of OTT streaming content.

The analysis covers content type, release years, genres,
countries, ratings, movie duration and TV show seasons.

These insights can help streaming platforms understand
content trends, identify growth opportunities and support
future content acquisition and investment decisions.
""")
```
